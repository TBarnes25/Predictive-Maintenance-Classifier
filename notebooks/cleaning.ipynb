```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd
import numpy as np
import math
import pickle
```

```python

  
# fetch dataset 
ai4i_2020_predictive_maintenance_dataset = fetch_ucirepo(id=601) 
# data (as pandas dataframes) 
x = ai4i_2020_predictive_maintenance_dataset.data.features 
y = ai4i_2020_predictive_maintenance_dataset.data.targets 
df = pd.concat([x,y], axis =1)

display(df.head(25))
df = df.drop(columns= ["TWF","HDF","PWF","OSF","RNF"])
display(df.head(25))

display(df.info())
  
```

```python
#Power in watts
df["Power"]  = (df["Torque"] * df["Rotational speed"] * (math.pi /30)).round(0)
df["Overstrain"] = (df["Torque"] * df["Tool wear"]).round(0)
df["Temperature difference"] = (df["Process temperature"] - df["Air temperature"]).round(0)
display(df.sample(30))
```

```python
#Normalising the type , will jsut detect it the overstrain exceeds 11,000
df["Type"] = df["Type"].replace({
    "L": 0,
    "M": 1,
    "H": 2
})
df  = df.astype({"Type":"int"})
display(df.head(15))
```

```python

df = df.rename(columns= lambda x: x.lower().replace(" ","_"))
display(df.head())
```

Power feature: Mechanical power produced by the motor. Too low or too high could indicate a fault indicated previsouly by PWF
Overstrain feature: Cummulativemechanical stress overtime, previously indicated by OSF
Temp difference: How effectively heat generated by the process is dissipating
